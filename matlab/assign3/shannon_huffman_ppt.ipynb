```python
#SAHNNON
import matplotlib.pyplot as plt
import networkx as nx

def draw_prefix_tree(code_dict, title="Prefix Code Tree"):
    """
    Takes a dictionary of characters and their binary codes and draws a tree.
    Example: code_dict = {'C': '0', 'A': '11', 'B': '100', 'D': '101'}
    """
    G = nx.DiGraph()
    G.add_node("Root")

    # Build the graph based on the binary codes
    for char, code in code_dict.items():
        current_node = "Root"
        path_so_far = ""

        for bit in code:
            next_node = path_so_far + bit
            G.add_node(next_node)
            # Add an edge with the bit label (0 or 1)
            G.add_edge(current_node, next_node, label=bit)
            current_node = next_node
            path_so_far = next_node

        # Rename the final leaf node to include the character
        leaf_label = f"{char}\n({code})"
        nx.relabel_nodes(G, {current_node: leaf_label}, copy=False)

    # Use graphviz for a strict top-down tree layout if available,
    # otherwise fallback to networkx's spring layout (requires 'pydot' installed)
    try:
        from networkx.drawing.nx_pydot import graphviz_layout
        pos = graphviz_layout(G, prog="dot")
    except ImportError:
        print("For a perfect top-down layout, install graphviz and pydot (pip install pydot). Using fallback layout.")
        pos = nx.spring_layout(G, seed=42)

    # 1. MASSIVE Canvas with High DPI for zooming in
    plt.figure(figsize=(30, 20), dpi=300)

    # Identify leaf nodes vs internal nodes
    out_degrees = dict(G.out_degree())
    leaves = [node for node in G.nodes() if out_degrees[node] == 0]
    internals = [node for node in G.nodes() if out_degrees[node] > 0]

    # 2. Shrink the node sizes so they don't overlap
    nx.draw_networkx_nodes(G, pos, nodelist=internals, node_size=600, node_color='lightblue', node_shape='o')
    nx.draw_networkx_nodes(G, pos, nodelist=leaves, node_size=1200, node_color='lightgreen', node_shape='s')

    # 3. Draw edges and force a much smaller font size for the node text
    nx.draw_networkx_edges(G, pos, arrows=False)
    nx.draw_networkx_labels(G, pos, font_size=8, font_weight="bold")

    # 4. Shrink the '0' and '1' edge labels
    edge_labels = nx.get_edge_attributes(G, 'label')
    nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=8, font_color='red')

    plt.title(title, fontsize=24, fontweight='bold')

    # 5. Tight layout removes wasted white space around the edges
    plt.tight_layout()
    plt.axis("off")

    # Save it directly to a file so you can open it in an image viewer and zoom in!
    plt.savefig("shannon_tree_high_res.png", format="png", bbox_inches="tight")
    plt.show()

# --- INPUT YOUR MATLAB RESULTS HERE ---
# Replace this dictionary with the Shannon or Huffman codes from your MATLAB terminal
my_matlab_codes = {
        ' ': '000',
        '.': '11110110',
        'K': '11111000',
        'M': '1101111',
        'a': '0100',
        'b': '11111010',
        'c': '10101',
        'd': '01110',
        'e': '0010',
        'f': '110011',
        'g': '1110001',
        'h': '10110',
        'i': '0101',
        'l': '10000',
        'm': '1110011',
        'n': '10010',
        'o': '10011',
        'p': '1110101',
        'q': '1110111',
        'r': '10111',
        's': '11000',
        't': '01100',
        'u': '110101',
        'v': '11111100',
        'w': '11111110',
        'x': '1111001',
        'y': '110110'
}

draw_prefix_tree(my_matlab_codes, title="My Assigned Shannon Coding Tree")
```

```python
#HUFFMAN
import matplotlib.pyplot as plt
import networkx as nx

def draw_huffman_tree(code_dict, title="My Assigned Huffman Coding Tree"):
    """
    Takes a dictionary of characters and their binary Huffman codes and draws a high-res tree.
    """
    G = nx.DiGraph()
    G.add_node("Root")

    # Build the graph based on the binary codes
    for char, code in code_dict.items():
        current_node = "Root"
        path_so_far = ""

        for bit in code:
            next_node = path_so_far + bit
            G.add_node(next_node)
            G.add_edge(current_node, next_node, label=bit)
            current_node = next_node
            path_so_far = next_node

        # Format the leaf label: Replace spaces with 'SPACE' for readability
        display_char = "SPACE" if char == " " else char
        leaf_label = f"{display_char}\n({code})"
        nx.relabel_nodes(G, {current_node: leaf_label}, copy=False)

    # Use graphviz for a strict top-down tree layout
    try:
        from networkx.drawing.nx_pydot import graphviz_layout
        pos = graphviz_layout(G, prog="dot")
    except ImportError:
        print("Please install graphviz and pydot for top-down layout (pip install pydot).")
        pos = nx.spring_layout(G, seed=42)

    # --- HIGH RESOLUTION SETTINGS ---
    plt.figure(figsize=(30, 20), dpi=300)

    out_degrees = dict(G.out_degree())
    leaves = [node for node in G.nodes() if out_degrees[node] == 0]
    internals = [node for node in G.nodes() if out_degrees[node] > 0]

    # Shrink nodes to prevent overlap in large trees
    nx.draw_networkx_nodes(G, pos, nodelist=internals, node_size=600, node_color='lightcoral', node_shape='o')
    nx.draw_networkx_nodes(G, pos, nodelist=leaves, node_size=1200, node_color='gold', node_shape='s')

    nx.draw_networkx_edges(G, pos, arrows=False)

    # Tiny font size to keep text inside the bubbles
    nx.draw_networkx_labels(G, pos, font_size=8, font_weight="bold")

    edge_labels = nx.get_edge_attributes(G, 'label')
    nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=8, font_color='red')

    plt.title(title, fontsize=24, fontweight='bold')

    plt.tight_layout()
    plt.axis("off")

    # Saves directly to your folder so you can zoom in!
    plt.savefig("huffman_tree_high_res.png", format="png", bbox_inches="tight")
    print("Success! Open 'huffman_tree_high_res.png' to view your tree.")
    plt.show()

# --- INPUT YOUR MATLAB RESULTS HERE ---
# Copy the [Huffman Code] column from your MATLAB terminal for your specific sentence
my_huffman_codes = {
    ' ': '001',
    '.': '1000011',
    'K': '1000010',
    'M': '100000',
    'a': '0001',
    'b': '01001001',
    'c': '01000',
    'd': '1110',
    'e': '101',
    'f': '000001',
    'g': '100011',
    'h': '01011',
    'i': '0110',
    'l': '1101',
    'm': '100010',
    'n': '1100',
    'o': '1111',
    'p': '100101',
    'q': '100100',
    'r': '01010',
    's': '00001',
    't': '0111',
    'u': '000000',
    'v': '01001000',
    'w': '0100101',
    'x': '010011',
    'y': '10011'
}

draw_huffman_tree(my_huffman_codes)
```
