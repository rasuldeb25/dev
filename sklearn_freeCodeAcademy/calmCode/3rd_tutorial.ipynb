```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler, QuantileTransformer
from sklearn.datasets import make_blobs
```

```python
# generate two clusters
X, y = make_blobs(
    n_samples=500,
    centers=[(400, 500), (1100, 400)],  # positive coordinates
    cluster_std=160,
    random_state=90
)

# ensure all values are positive
X = np.abs(X)

df = pd.DataFrame({
    "x": X[:,0],
    "y": X[:,1],
    "z": ["a" if label == 0 else "b" for label in y]
})

df.to_csv("drawndata1.csv", index=False)
df.head()
```
