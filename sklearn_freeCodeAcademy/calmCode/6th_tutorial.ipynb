```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("C:/dev/sklearn_freeCodeAcademy/calmCode/creditcard.csv")[:80_000]
df.head(3)
```

```python
X = df.drop(columns=["Time", "Amount", "Class"]).values
y = df["Class"].values
f"Shapes of X={X.shape} y={y.shape}, Fraud Cases={y.sum()}"
```

```python
from sklearn.linear_model import LogisticRegression

mod = LogisticRegression(class_weight={0:1,1:2}, max_iter=1000)
mod.fit(X,y).predict(X).sum()
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import precision_score, recall_score, make_scorer

grid = GridSearchCV(
    estimator=LogisticRegression(max_iter=1000),
    param_grid={"class_weight": [{0:1, 1:v} for v in np.linspace(1, 20, 30)]},
    scoring={"precision": make_scorer(precision_score), "recall": make_scorer(recall_score)},
    refit="precision",
    return_train_score=True,
    cv=10,
    n_jobs=-1
)

grid.fit(X,y)
```

```python
# pd.DataFrame(grid.cv_results_)
```

```python
plt.figure(figsize=(12,4))
df = pd.DataFrame(grid.cv_results_)
for score in ["mean_test_recall", "mean_test_precision"]:
    plt.plot([_[1] for _ in df["param_class_weight"]],
             df[score],
             label=score)
plt.legend();
```

```python
plt.figure(figsize=(12,4))
df = pd.DataFrame(grid.cv_results_)
for score in ["mean_train_recall", "mean_train_precision"]:
    plt.scatter([_[1] for _ in df["param_class_weight"]],
             y=df[score.replace("test","train")],
             label=score)
plt.legend();
```

```python
from sklearn.model_selection import GridSearchCV

def outlier_precision(mod, X,y):
    preds = mod.predict(X)
    return precision_score(y, np.where(preds == -1,1,0))
def outlier_recall(mod,X,y):
    preds = mod.predict(X)
    return recall_score(y, np.where(preds == -1,1,0))


grid = GridSearchCV(
    estimator=IsolationForest(),
    param_grid={"contamination": np.linspace(0.001, 0.02,10) },
    scoring={"precision": outlier_precision,
             "recall": outlier_recall},
    refit="precision",
    cv=5,
    n_jobs=-1
)

grid.fit(X,y);
```

```python
plt.figure(figsize=(12,4))
df_results = pd.DataFrame(grid.cv_results_)
for score in ["mean_test_recall", "mean_test_precision"]:
    plt.plot(df_results["param_contamination"],
             df_results[score],
             label=score)
plt.legend();
```
