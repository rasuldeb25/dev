```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
medical_df = pd.read_csv("C:/dev/sklearn_freeCodeAcademy/full_course_Aakash_N_S/insurance.csv", encoding="utf-8-sig", low_memory=False)
```

```python
non_smoker_df = medical_df[medical_df.smoker == "no"]
```

```python
def rmse(targets, predictions):
    return np.sqrt(np.mean(np.square(targets - predictions)))
```

```python
#creating inputs and targets
inputs, targets = non_smoker_df[["age"]], non_smoker_df["charges"]
#creating and training the model
model = LinearRegression().fit(inputs,  targets)
#predictions
predictions  = model.predict(inputs)
#computing the loss function and evaluating the model
loss = rmse(targets, predictions)
print("loss: ", loss)
```

```python
#adding more features than only one
inputs, targets = non_smoker_df[["age", "bmi"]], non_smoker_df["charges"]
model = LinearRegression().fit(inputs, targets)
predictions = model.predict(inputs)
predictions = model.predict(inputs)
loss = rmse(targets, predictions)
print("loss: ", loss)
```
